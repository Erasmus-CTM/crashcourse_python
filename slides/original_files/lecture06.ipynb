## NumPy arrays &hyphen; vectors
### The array type

Lists are almost like vectors but operations on lists are **not** linear algebra operations.

```python
L1 = [1, 2]
L2 = [3, 4]
print("L1 + L2 =", L1 + L2)
print("3*L1 =", 3*L1)
```

#### Definition
An **array** represents a vector or a matrix in linear algebra. It is often initialised from a list or another vector. Operations `+`, `*`, `/`, `-` are all **elementwise**. `dot` or the `@` operator are used for the scalar product.

## Vector usage

#### Example

```python
from numpy import array, dot, sqrt, inf
from scipy.linalg import norm

vec = array([3, -4]) # a vector in the plane

print("vec =", vec)
print("2*vec =", 2*vec)
print("vec*vec =", vec*vec)
print("vec/2 =", vec/2) 
print("vec/vec =", vec/vec) 

# TODO: explain norms 
print("||vec|| =", norm(vec, ord=2))
print("|vec| = ", sqrt(dot(vec,vec))) # same as norm(vec, ord=2)

# TODO: explain
print("scalar product =", dot(vec, vec))
print("scalar product =", vec @ vec)
```

## Vectors are not lists!

Operations are not the same:

- Operations `+` and `*` are different
- More operations are defined: `-`, `/`
- Numpy functions act elementwise on vectors: `sin`, `exp`, `sqrt`, etc.
- Scalar product with `dot` or `@`
- Vectors have a **fixed size**: no `append` method!

## Vectors are not lists! (cont)

Two names are bound to the same list:

```python
L1 = [1, 2]
L2 = L1
L1[0] = 10
print(L2)
L2[0] = 4
print(L1)
```

To make an actual copy, you can use the `copy` method, or make a slice.

```python
L1 = [1, 2] # compare with with numpy array (next silde)
L2 = L1 
L3 = L1.copy()
L4 = L1[:]  # a slice of the entire list
L1[0] = 10  # what happens here? 
print("L2 =", L2)
print("L3 =", L3)
print("L4 =", L4)
```

## Vectors are not lists! (cont)

- Vector slices, on the other hand, are **views**:

```python
L1 = [1, 2, 3]

v1 = array([1, 2, 3])

L2 = L1[:2]
L2[0] = 10
print("now L1 = ", L1)

v3 = v1[:2]
#v3[0] = 10
print("now v1 = ",v1)

v2 = v1[:2]
v2[0] = 10
print("now v1 = ", v1)

v3 = v1.copy()[:2]
v3[0] = 10
print("now v3 = ", v3)
```

## Accessing matrix entries

Matrix elements are accessed with **two** indices.

```python
M = array([[1., 2.], [3., 4.]])

print(M[0, 0])  # first row, first column

print(M[-1, 0]) # last row, first column

# Don't do this:
print(M[0][0])
print(M[0])
```

Compare this to accessing list elements.

```python
L = [[1., 2.], [3., 4.]]

print(L[0][0])

print(L[-1][0])
```

## Transpose

The transpose of a matrix $A$ is a matrix $B$ such that

$$B_{ij} = A_{ji}.$$

By transposing a matrix you switch the two shape elements.

```python
A = array([[1, 2, 3],
           [4, 5, 6]])
print("A.shape =", A.shape)
print(A[1,2])

B = A.T  # A transpose
print("B.shape =", B.shape, '\n')

#B[0,0] = 11
print("A =")
print(A, '\n')
print("B =")
print(B)
```

## Matrix multiplication

We can use the `dot` function for matrix multiplication. We can also use the `@` operator.

If $A$ is a $n\times m$ matrix and $B$ a $m \times p$ matrix, then $AB$ is a $n\times p$ matrix.

```python
from numpy import dot
A = array([[1, 0, 0],        # n = 2, m = 3
           [2, -1, 3]])
B = array([[2, 1, 0, 4],     # m = 3, p = 4
           [1, 1, -1, 0],
           [0, 1, 2, 2]])

  

AB = dot(A, B)
print(AB)

# same as 
AB = A @ B
print(AB)

# BA = dot(B, A)  # error
```

## Matrix vector multiplication

A matrix can be multipled with a vector using `dot` or the `@` operator. 
The multiplication is matrix multiplication, as if the vector was a column or row matrix. The result is a vector.

$$ 
\left(
\begin{array}{rr}
1 & 2
\end{array}
\right)
\left(
\begin{array}{rr}
1 & 0 \\
4 & -1
\end{array}
\right) 
= 
\left(
\begin{array}{rr}
9 & -2
\end{array}
\right)
$$

$$
\left(
\begin{array}{rr}
1 & 0 \\
4 & -1
\end{array}
\right) 
\left(
\begin{array}{rr}
1 \\
2
\end{array}
\right)
= 
\left(
\begin{array}{rr}
1 \\
2
\end{array}
\right)
$$

```python
A = array([[1, 0],
           [4, -1]])
v = array([1, 2])

vA = dot(v, A)
vA = v @ A
Av = dot(A, v)

print(vA)
print(Av)
print("vA.shape =", vA.shape, " Av.shape = ", Av.shape)
```

## Dot product and `@` operator


| vector vector | $s = \sum_i x_i y_i$ | `s = x @ y` or `s = dot(x,y)` | scalar product |
|:---|:---|:---|:---|
| matrix matrix | $C_{ij} = \sum_k A_{ik}B_{kj}$ | `C = A @ B` or `C = dot(A, B)` | matrix multiplication |
| matrix vector | $y_i = \sum_j A_{ij}x_j$ | `y = A @ x` or `y = dot(A, x)` | matrix multiplication, treating vector as column matrix |
| vector matrix | $y_j = \sum_i x_i A_{ij}$ | `y = x @ A` or `y = dot(x, A)` | matrix multiplication, treating vector as row matrix |

## Linear transformations ‐ rotation matrix in the plane

### Definition

A **linear transformation** in the plane is a function $f:\mathbb{R}^2 \mapsto \mathbb{R}^2$  such that


$f(a\mathbf{u} + b\mathbf{v}) = af
(\mathbf{u}) + bf(\mathbf{v}) \text{ for all }a, b\in\mathbb{R} \text{ and } \mathbf{u}, \mathbf{v} \in \mathbb{R}^2$

## Matrices and linear transformations

Let $x= (x_1, x_2) \in \mathbb{R}^2$ and $y= (y_1, y_2) \in \mathbb{R}^2$. Let $y = f(x)$ be the function defined by

$$
\begin{cases}
y_1 = ax_1 + bx_2 \\
y_2 = cx_1 + dx_2
\end{cases}
$$

for some numbers $a, b, c, d$. Then matrix multiplication can be used to describe $f$.
Let 

$$A = 
\left(
\begin{array}{rr}
a & b \\
c & d
\end{array}
\right),
x = 
\left(
\begin{array}{rr}
x_1 \\
x_2
\end{array}
\right),
y = 
\left(
\begin{array}{rr}
y_1 \\
y_2
\end{array}
\right).
\text{ Then }
y = Ax, \text{ i.e. }
\left(
\begin{array}{rr}
y_1 \\
y_2
\end{array}
\right)
=
\left(
\begin{array}{rr}
a & b \\
c & d
\end{array}
\right)
\left(
\begin{array}{rr}
x_1 \\
x_2
\end{array}
\right)
$$

$A$ is a **transformation matrix** of $f$, $y = f(x) = Ax$

## Finding the transformation matrix of a rotation in the plane

Let $\mathbf{e_1}, \mathbf{e_2}$ be two basis vectors and let $A$ be a transformation matrix that rotates a vector $\mathbf{u} = x\mathbf{e_1} + y\mathbf{e_2}$ by an angle $\alpha $. To find $A$ it suffices to consider the basis vectors.

![rotated basis](http://cmc.education/slides/notebookImages/rotatedBasis.svg)

Now $ A\mathbf{u} = xA\mathbf{e_1} + yA\mathbf{e_1} = x\mathbf{e_1}' + y\mathbf{e_2}'$

### Rotate a point

```python
from matplotlib.pyplot import *
from numpy import *
%matplotlib inline
alpha = pi/2
A = array([[cos(alpha), -sin(alpha)],
           [sin(alpha), cos(alpha)]])
x = 3
y = 3
P = array([x, y]) # a vector
AP = dot(A, P)    # a vector
AP = A @ P 
plot(x, y, 'bo')
xlim(-5, 5)
ylim(-5, 5)
grid()
plot(*AP, 'ro')  # unpack to get x and y
```

### Rotate many points

**5min, let's go!**

```python
alpha = pi/2
A = array([[cos(alpha), -sin(alpha)],
           [sin(alpha), cos(alpha)]])
x = linspace(-10, 10, 200)
y = sin(3*x)
M = array([x, y]) # 2 times 200 matrix
AM = dot(A, M)    # 2 times 200 matrix
plot(x, y)
grid()
plot(*AM)  # unpack to get a x-list and a y-list
```
