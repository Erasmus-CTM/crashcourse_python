# Unpacking lists and tuples

### Examples


````python
lst1 = [1, 2, 3, 4, 5]
a, b, c, d, e = lst1
print(b)
````

If we're only interested in the first two elements, we can use an asterisk.

````python
a, b, *the_rest = lst1
print(a)
print(b)
print(the_rest)
````

## Packing and unpacking in assignments

The numbers 1 and 2 are packed into a tuple.

The tuple is unpacked in the assignment.

````python
a, b = 1, 2
print(a)
print(b)
````

This is useful for making swaps

````python
a, b = b, a
print(a)
print(b)
````

## Unpacking a list of one single element

We will need this later.

````python
lst2 = [15]

a, = lst2
print(a)
````

The comma indicates that this is an `unpacking`.

## Packing using asterisk operators (* and **)

### Asterisks in a function definition
The words args and kwargs are often used but the parameters may have any valid names. 

````python
def myfunc(*args, **kwargs):
    print("Printing arguments.")
    print(type(args))
    for arg in args:
        print(arg)
        
    print("Printing keyword arguments.")
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f" {key} : {value} ")
        
myfunc(1, 2, 3, nr1 = 4, nr2 = 5)
````

## Unpacking using asterisk operators (* and **)

### Asterisks when calling a function

````python
def myfunc(*args, **kwargs):
    print("Printing arguments.")
    for arg in args:
        print(arg)
        
    print("Printing keyword arguments.")
    for key, value in kwargs.items():
        print(f" {key} : {value} ")
        
        
lst = [1, 2, 3]
print(lst[1])
d = {"nr1": 4, "nr2": 5}
print(d["nr1"])

myfunc(*lst, **d)  # what would happen if all asterisks were removed?
````

## Asterisk operators to handle a variable number of arguments

### Example using the function len

````python
def myfunc(*numbers):
    '''
    This function should be called with one or two arguments that are numbers.
    '''
    if len(numbers) == 1:
        print("Called with one argument.")
    else:
        print("Called with more than one arguments.")
        
myfunc(5, 7, 8)
````

# More about plotting

## State machine or object-oriented programming (OOP)

`matplotlib.pyplot` can be used with command-like functions.

* Various states are preserved.
* A state holds information about `current figure` and `current axes`.
* Under the hood, OOP is used.

`matplotlib.pyplot` can be also be used for OOP.
* Instead of functions, methods of objects are used.

For event driven programs, OOP **should** be used.


### First some imports

````python
from matplotlib.pyplot import *
from numpy import *
%matplotlib inline     
# this is a magic command for static plots, we will do interactive plots later
````

## State machine - two figures
The function call `plot` can be viewed as a command. The same command results in different outcomes depending on the state.

````python
x1 = linspace(-2*pi, 2*pi, 100)
x2 = linspace(0, 2, 10)

plot(x1, sin(x1), 'r', label='sin(x)') # a figure is created
plot(x1, cos(x1), 'b--', label='cos(x)')
xlabel('x')
legend()
title('My first figure')
 
figure() # a new figure is created 

plot(x2, sqrt(x2), 'go')
plot(x2, x2**2, 'y^')
title('My second figure')
show()  # show is needed in editors that don't have an interactive console
````

## State machine - two subplots in one figure
The plots are made in a subplot of the <span class=alert>current axes</span> in the <span class=alert>current figure</span>.

````python
x1 = linspace(-2*pi, 2*pi, 100)
x2 = linspace(0, 2, 10)

subplot(211) # the first subplot using a 2 rows and 1 column grid
#abtight_layout()
plot(x1, sin(x1), 'r')
plot(x1, cos(x1), 'b--')
title('My first subplot')
 
subplot(212) # the second subplot using a 2 rows and 1 column grid
plot(x2, sqrt(x2), 'go')
plot(x2, x2**2, 'y^') 
subplots_adjust(left=None, bottom=None, right=None, top=None, wspace=0.2, hspace=0.5)

title('My second subplot') 
show()
````

## Axes objects &hyphen; Ticks and ticklabels

````python
fig, ax = subplots(figsize=(8, 3))
x =linspace(-2*pi, 2*pi, 100)
ax.plot(x, sin(x))

ax.set_xticks([-pi, 0, pi])
ax.set_xticklabels(['$-\pi$', '$0$', '$\pi$'], fontsize = 16)  # using LaTeX

# set minor ticks
ax.set_xticks([-pi*1.5, -pi*0.5, pi*0.5, pi*1.5], minor = True)

# use empty array to hide ticks
ax.set_yticks([])
````

## Axes objects &hyphen; Spines

By default four **spines** are rendered at the boundary of a plot. We can change the position of spines, and hide them.

````python
fig, ax = subplots(figsize=(8, 3))
x =linspace(-2*pi, 2*pi, 100)
ax.plot(x, sin(x))

ax.spines['left'].set_position('zero')
ax.spines['right'].set_color('none')      # hide
ax.spines['bottom'].set_position('zero')
ax.spines['top'].set_color('none')        # hide
````

## Annotation

The arrow is given properties using a dictionary. (See the textbook for more examples)

````python
fig, ax = subplots(figsize = (8, 3))
x =linspace(-2*pi, 2*pi, 100)
ax.plot(x, sin(x))

# we need space above the curve
ax.set_ylim([-1.2, 2])

ax.annotate("local max", xy = (pi/2, 1), xytext = (pi, 1.8),
            arrowprops = {'facecolor': 'green', 'shrink': 0.1, 'width': 3})
````

## Fill the area between curves

````python
fig, (ax1, ax2) = subplots(nrows = 1, ncols = 2, figsize = (15, 3))
x =linspace(-2*pi, 2*pi, 100)

y1 = sin(x)
y2 = cos(x)

ax1.plot(x, y1, 'r')
ax1.plot(x, y2, 'b')
ax1.fill_between(x, y1, y2, color = 'green', alpha = 0.9)

# specify fill condition using where
ax2.plot(x, y1, 'r')
ax2.plot(x, y2, 'b')
ax2.fill_between(x, y1, y2, where = y1 > y2, color = 'red', alpha = 0.3)
````

## RTFM 

Please take a look at the matplotlib gallery. There are a lot of examples to start from: 

[https://matplotlib.org/stable/gallery/](https://matplotlib.org/stable/gallery/)

````python
import matplotlib.pyplot as plt
import numpy as np


def koch_snowflake(order, scale=10):
    """
    Return two lists x, y of point coordinates of the Koch snowflake.

    Parameters
    ----------
    order : int
        The recursion depth.
    scale : float
        The extent of the snowflake (edge length of the base triangle).
    """
    def _koch_snowflake_complex(order):
        if order == 0:
            # initial triangle
            angles = np.array([0, 120, 240]) + 90
            return scale / np.sqrt(3) * np.exp(np.deg2rad(angles) * 1j)
        else:
            ZR = 0.5 - 0.5j * np.sqrt(3) / 3

            p1 = _koch_snowflake_complex(order - 1)  # start points
            p2 = np.roll(p1, shift=-1)  # end points
            dp = p2 - p1  # connection vectors

            new_points = np.empty(len(p1) * 4, dtype=np.complex128)
            new_points[::4] = p1
            new_points[1::4] = p1 + dp / 3
            new_points[2::4] = p1 + dp * ZR
            new_points[3::4] = p1 + dp / 3 * 2
            return new_points

    points = _koch_snowflake_complex(order)
    x, y = points.real, points.imag
    return x, y
````

````python
x, y = koch_snowflake(order=5)

plt.figure(figsize=(8, 8))
plt.axis('equal')
plt.fill(x, y)
plt.show()
````
